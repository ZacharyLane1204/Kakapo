```python
import os
import requests
from bs4 import BeautifulSoup

from tqdm import tqdm
```

```python
base_url = "https://archive.stsci.edu/pub/k2/download_scripts/target-pixel-files/c16/"

# Set to track unique, non-empty lines
unique_lines = set()

# Fetch the directory page
response = requests.get(base_url)
response.raise_for_status()
soup = BeautifulSoup(response.text, "html.parser")

# Loop through all links
for link in soup.find_all("a"):
    href = link.get("href")
    if href and href.endswith(".txt"):
        txt_url = base_url + href
        print(f"Processing {txt_url}")
        txt_response = requests.get(txt_url)
        txt_response.raise_for_status()
        
        # Read and filter lines
        for line in txt_response.text.splitlines():
            stripped = line.strip()
            if stripped and stripped not in unique_lines:
                unique_lines.add(stripped)

# Write to the output file
with open("combined_output.txt", "w") as output_file:
    for line in sorted(unique_lines):
        output_file.write(line + "\n")
```
